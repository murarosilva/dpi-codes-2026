```python
import os
import pandas as pd
import PyPDF2
import re

# Define the directory containing the PDF files
pdf_directory = 'PDFs_May_2025'

# Initialize an empty list to store the data
data = []

# Function to clean text by removing illegal characters
def clean_text(text):
    return re.sub(r'[^\x00-\x7F]+', '', text)

# Iterate over all files in the directory
for filename in os.listdir(pdf_directory):
    if filename.endswith('.pdf'):
        file_path = os.path.join(pdf_directory, filename)
        
        try:
            # Open the PDF file
            with open(file_path, 'rb') as file:
                reader = PyPDF2.PdfReader(file)
                content = ''
                
                # Check if the PDF is encrypted
                if reader.is_encrypted:
                    try:
                        reader.decrypt('')
                    except:
                        print(f"Error decrypting {filename}, skipping this file.")
                        continue
                
                # Extract text from each page
                for page_num in range(len(reader.pages)):
                    page = reader.pages[page_num]
                    content += page.extract_text()
                
                # Clean the content
                content = clean_text(content)
                
                # Append the content and filename to the data list
                data.append({'document_name': filename, 'content': content})
                print(f"Successful reading {filename}, going to next file.")
        
        except PyPDF2.errors.PdfReadError:
            print(f"Error reading {filename}, skipping this file.")

# Create a DataFrame from the data list
df = pd.DataFrame(data)
```
