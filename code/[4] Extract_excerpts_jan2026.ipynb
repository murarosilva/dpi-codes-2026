```python
keywords = ["digital public infrastructure", "public digital infrastructure", "dpi", "digital public infrastructures"]
# Count occurrences of each keyword in the 'content' column and sum them
def count_dpi_mentions(text):
    return sum(len(re.findall(rf'\b{re.escape(keyword)}\b', text, re.IGNORECASE)) for keyword in keywords)

pdfs_content_df['DPI_mentions'] = pdfs_content_df['content'].apply(count_dpi_mentions)
```

# Extract Excerpts from text

# Extract Excerpts 3 - all 3 sequential phrases (similar to Timo)

```python
# Function to extract excerpts of 3 sequential sentences
def extract_sequential_3sentences(content):
    # Split the content into sentences
    sentences = re.split(r'(?<!\w\.\w.)(?<![A-Z][a-z]\.)(?<=\.|\?)\s', content)
    excerpts = []
    # Combine three sequential sentences
    for i in range(len(sentences) - 2):
        excerpt = ' '.join(sentences[i:i+3])
        excerpts.append(excerpt.strip())
    return excerpts

# Apply the function to extract excerpts and create a new dataframe
pdfs_content_df['sequential_sentences_excerpts'] = pdfs_content_df['content'].apply(extract_sequential_3sentences)
sequential_3excerpts_df = pdfs_content_df[['document_name', 'sequential_sentences_excerpts']].explode('sequential_sentences_excerpts').dropna().reset_index(drop=True)
```
